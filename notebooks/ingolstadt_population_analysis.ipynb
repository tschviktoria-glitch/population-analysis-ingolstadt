### Ziel des Projekts <a id="ziel"></a>
Ziel des Projekts ist die Analyse der Bevölkerungsstruktur in Ingolstadt hinsichtlich der Herkunft, der Altersstruktur sowie der räumlichen Verteilung auf die Stadtbezirke. Dabei wird untersucht, wie sich die verschiedenen Herkunftsgruppen räumlich und zeitlich verteilen und inwiefern sich ihre Altersstruktur von der der Gesamtbevölkerung unterscheidet.

Gerade in Städten wie Ingolstadt ist es relevant zu verstehen, wie sich Zuwanderung auf die demografische Zusammensetzung auswirkt und ob sie zu einer „Verjüngung“ der Bevölkerung beiträgt.

## Theoretischer Hintergrund <a id="theorie"></a>

### Migration <a id="migration"></a>
Unter Migration versteht man die dauerhafte oder vorübergehende Verlagerung des Wohnsitzes über eine Landesgrenze hinweg. Personen mit Migrationshintergrund sind laut Definition des Statistischen Bundesamtes alle, die selbst oder deren Eltern nicht mit deutscher Staatsangehörigkeit geboren wurden.
### Altersstruktur der Bevölkerung <a id="alter"></a>
Die Altersstruktur beschreibt die Verteilung der Bevölkerung auf verschiedene Altersgruppen. 
Die Altersstruktur einer Bevölkerung beeinflusst langfristig den Arbeitsmarkt, die Nachfrage nach Bildungs- und Gesundheitseinrichtungen sowie kommunale Planungsprozesse. 
In Deutschland führt der demografische Wandel zu einer zunehmenden Alterung der Bevölkerung, was sich negativ auf das Arbeitskräfteangebot und die Sozialsysteme auswirken kann.
### Einfluss der Migration auf die Demografie <a id="einfluss"></a>
Migration kann den Alterungsprozess einer Bevölkerung verlangsamen, wenn der Anteil junger Menschen unter den Zuwanderern höher ist als in der einheimischen Bevölkerung. Dadurch kann sie einen Beitrag zur „Verjüngung“ der Altersstruktur leisten.

#### Auslaender nach Staatengruppen-Unterbezirksebene in Ingolstadt

Wir überprüfen, ob die Spaltennamen und Datentypen mit dem ersten Tabellenblatt „Afrika“ übereinstimmen, damit dieses im weiteren Verlauf als Grundlage verwendet werden kann.

```python
# Referenz: Spalten + Datentypen
reference_dtypes = df4[list(df4.keys())[0]].dtypes

for name, df in df4.items():
    if not df.dtypes.equals(reference_dtypes):
        print(f"\n❌ {name} weist Abweichungen bei Datentypen auf")
        
        # Unterschiede anzeigen
        for col in df.columns:
            if col in reference_dtypes:
                if df[col].dtype != reference_dtypes[col]:
                    print(f"Spalte '{col}': {df[col].dtype} statt {reference_dtypes[col]}")
            else:
                print(f"Spalte '{col}' existiert nicht in der Referenz")
    else:
        print(f"✅ {name} hat identische Datentypen")
```

Die Überprüfung zeigt, dass alle Tabellenblätter innerhalb dieses Datensatzes identische Spaltennamen und Datentypen aufweisen.Daher kann das Tabellenblatt „Afrika“ als Referenzstruktur für die weiteren Analysen verwendet werden.

Es wurde entschieden, für alle Datensätze eine zusätzliche Spalte „Bezirk_standard“ zu erstellen. Diese enthält einheitliche, standardisierte Bezirksbezeichnungen gemäß der zuvor definierten Referenzliste.

Dabei bleiben die ursprünglichen Spalten unverändert erhalten, während die neue Spalte eine konsistente Grundlage für weitere Analysen und Vergleiche schafft.

```python
def clean_bezirk(x):
    if isinstance(x, str):
        x = x.strip()
        x = re.sub(r"^\d+\s*-\s*", "", x)
        x = re.sub(r"\.\-", "-", x)
        x = re.sub(r"\s+", " ", x)
    return x
```

```python
for df in [df1, df3]:
    df["Bezirk_standart"] = df["Bezirk"].apply(clean_bezirk)

for key, df in df5.items():
    df["Bezirk_standart"] = df["Bezirk"].apply(clean_bezirk)
```

```python
for df in [df2]:
    df["Bezirk_standart"] = df["Gebiet"].apply(clean_bezirk)

for key, df in df4.items():
    df["Bezirk_standart"] = df["Gebiet"].apply(clean_bezirk)
```

```python
all_bezirke = set()

for df in [df1, df2, df3] + list(df4.values()) + list(df5.values()):
    all_bezirke.update(df["Bezirk_standart"].dropna().unique())

print(sorted(all_bezirke))
```

### Zusammenführung der Datensätze <a id="merge"></a>

#### Definition der Schlüsselvariablen (Keys)

Die Datenzusammenführung erfolgt anhand der beiden Schlüsselvariablen: Bezirk und Jahr.

In allen analysierten Dateien sind Daten für den Zeitraum von 2013 bis 2025 enthalten. Ursprünglich umfasste jede Datei Informationen zu 12 Bezirken. In einer der Dateien wurde ein zusätzlicher Bezirk mit der Bezeichnung "unbekannter Bezirk" hinzugefügt, um fehlende Daten zu ergänzen. Um eine korrekte Datenzusammenführung und die Wahrung der Datenintegrität zu gewährleisten, ist es erforderlich, vor der Konsolidierungsphase eine Zeile mit dem "unbekannten Bezirk" in alle anderen Dateien einzufügen.

Schlüsselvariablen:
- Bezirk (Name des Bezirks)
- Jahr

```python
def add_unknown_districts_per_file(dataframes, years=range(2013, 2026)):
    """
    Fügt Zeilen mit 'Unbekannter Bezirk' zu jedem DataFrame hinzu,
    wobei die eindeutigen Werte aus jedem DataFrame verwendet werden.

    Args:
        dataframes (list): Liste der DataFrames
        years (range): Jahresbereich (Standard: 2013-2025)

    Returns:
        list: Aktualisierte DataFrames
    """
    if not dataframes:
        raise ValueError("Die DataFrame-Liste darf nicht leer sein!")

    updated_dfs = []

    for df in dataframes:
        # 1. Entferne alle vorhandenen Zeilen mit 'Unbekannter Bezirk'
        df_clean = df[df['Bezirk'] != 'Unbekannter Bezirk'].copy()

        # 2. Hole eindeutige Werte für diesen DataFrame
        einheit_values = df_clean['Einheit'].unique()
        bezeichnung_values = df_clean['Bezeichnung'].unique()

        # Verwende den ersten eindeutigen Wert (alternativ könnte man auch den häufigsten wählen)
        einheit_wert = einheit_values[0] if len(einheit_values) > 0 else 'Anzahl'
        bezeichnung_wert = bezeichnung_values[0] if len(bezeichnung_values) > 0 else 'Ausländer'

        # 3. Erstelle Zeilen für 'Unbekannter Bezirk'
        unknown_rows = []
        for jahr in years:
            unknown_row = {
                'Bezirksnummer': -1,
                'Bezirk': 'Unbekannter Bezirk',
                'Gebiet': 'Unbekannt',
                'Einheit': einheit_wert,
                'Bezeichnung': bezeichnung_wert,
                'Bezirk_standard': 'Unbekannter Bezirk',
                'Wert': 0,
                'Jahr': jahr
            }
            unknown_rows.append(unknown_row)

        df_unknown = pd.DataFrame(unknown_rows)

        # 4. Kombiniere mit dem ursprünglichen DataFrame
        df_updated = pd.concat([df_clean, df_unknown], ignore_index=True)
        updated_dfs.append(df_updated)

    # 5. Gib die Ergebnisse aus
    print("\nFunktionsergebnis:")
    for i, df in enumerate(updated_dfs, 1):
        print(f"\nDataFrame {i}:")
        print(f"- Gesamtzahl der Zeilen: {len(df)}")
        print(f"- Eindeutige Werte 'Bezeichnung': {df['Bezeichnung'].unique()}")
        print(f"- Eindeutige Werte 'Einheit': {df['Einheit'].unique()}")
        print(f"- Beispielzeilen mit 'Unbekannter Bezirk':")
        print(df[df['Bezirk'] == 'Unbekannter Bezirk'].head())

    return updated_dfs

# Beispielaufruf
dataframes = [df1_long, df2_long, df3_long]
updated_dfs = add_unknown_districts_per_file(dataframes)

# Aktualisiere die Variablen
df1_long, df2_long, df3_long = updated_dfs
```

#### Alter

Die Altersangaben wurden zu größeren Gruppen zusammengefasst (0–17, 18–34, 35–49, 50–64, 65+), um die Analyse zu vereinfachen.

```python
def map_age(x):
    if x in [
        'von 0 bis unter 3 Jahren',
        'von 3 bis unter 6 Jahren',
        'von 6 bis unter 10 Jahren',
        'von 10 bis unter 15 Jahren',
        'von 15 bis unter 18 Jahren'
    ]:
        return "0–17"

    elif x in ['von 18 bis unter 25 Jahren', 'von 25 bis unter 30 Jahren']:
        return "18–34"

    elif x in ['von 30 bis unter 45 Jahren']:
        return "35–49"

    elif x in ['von 45 bis unter 60 Jahren', 'von 60 bis unter 65 Jahren']:
        return "50–64"

    elif x in [
        'von 65 bis unter 75 Jahren',
        'von 75 bis unter 85 Jahren',
        'ab 85 Jahren und älter'
    ]:
        return "65+"

    return None
```

```python
df_all["Alter_gruppe"] = df_all["Bezeichnung"].apply(map_age)
```

#### Herkunft

Die Herkunft wurde in übergeordnete Regionen aggregiert (z. B. EU, Nicht-EU Europa, Asien, Afrika), um die Vielzahl an Einzelkategorien zu reduzieren.

```python
def map_origin(x):
    if x in ["Türkei",
             'Ehemaliges Jugoslawien (ohne Slowenien u. Kroatien)*']:
        return "Europa(Türkei,Ehemaliges Jugoslawien, Nicht-EU)"

    elif "EU" in x or x in [
        'Griechenland',
        'Italien, Vatikan, San Marino',
        'Spanien, Portugal, Malta, Zypern, Andorra',
        'Nord- und Westeuropäische EU*'
    ]:
        return "EU"

    elif x == "Afrika":
        return "Afrika"

    elif x == "Amerika":
        return "Amerika"

    elif "Asiatische" in x or "Sowjetunion asiatisch" in x:
        return "Asien"

    elif "Sowjetunion europäisch" in x or \
         x == 'Nord- und Westeuropa, Nicht-EU*':
        return "Europa (inkl. ehemalige Sowjetunion, Nicht-EU)"

    elif "Australien" in x:
        return "Ozeanien"

    return None
```

```python
df_all["Herkunft_gruppe"] = None
df_all.loc[df_all["Typ"] == "Herkunft", "Herkunft_gruppe"] = \
df_all.loc[df_all["Typ"] == "Herkunft", "Bezeichnung"].apply(map_origin)
```

### Zusammenhang zwischen Alter und Migration <a id="zusammenhang"></a> 

Zu Beginn dieses Abschnitts wird geprüft, ob die Daten einer Normalverteilung folgen, um das geeignete statistische Verfahren für die weiteren Analysen auszuwählen.

```python
# =====================================================================
# 0. DATENBEREINIGUNG UND ERSTELLUNG DER PIVOT-TABELLEN
# =====================================================================

# Bereinigung der Daten von leeren oder unbekannten Stadtbezirken
df_clean = df_all[
    df_all['Bezirk_standart'].notna() & 
    (df_all['Bezirk_standart'] != 'Unbekannter Bezirk')
].copy()

# Aufteilung des Datensatzes nach Merkmalstyp (Migrationsstatus vs. Altersstruktur)
df_status = df_clean[df_clean['Typ'] == 'Status']
df_age = df_clean[df_clean['Typ'] == 'Alter']

# Erstellung der Pivot-Tabelle für den MIGRATIONSSTATUS
index_cols = ['Bezirk_standart', 'Jahr'] if 'Jahr' in df_clean.columns else 'Bezirk_standart'

pivot_status = df_status.pivot_table(
    index=index_cols, 
    columns='Bezeichnung', 
    values='Wert', 
    aggfunc='sum'
)

# Erstellung der Pivot-Tabelle für die ALTERSSTRUKTUR
pivot_alter = df_age.pivot_table(
    index=index_cols, 
    columns='Alter_gruppe', 
    values='Wert', 
    aggfunc='sum'
)


# =====================================================================
# 1. BERECHNUNG DER RELATIVEN INDIKATOREN (ANTEILE IN %)
# =====================================================================

# Berechnung der exakten Gesamtbevölkerung des Stadtbezirks (Nenner)
gesamt_population = pivot_status[[
    'Ausländer', 
    'Deutsche mit Migrationshintergrund', 
    'Deutsche ohne Migrationshintergrund'
]].sum(axis=1)

# Berechnung der Zielvariable (Migrationsanteil in %)
migration_anteil = (
    (pivot_status['Ausländer'] + pivot_status['Deutsche mit Migrationshintergrund']) 
    / gesamt_population
) * 100

# Definition der exakten Spaltennamen gemeldet aus dem Mapping
col_0_17 = '0–17'
col_18_34 = '18–34'
col_65_plus = '65+'

# Berechnung der prozentualen Anteile für die drei Altersgruppen
alter_0_17_anteil = (pivot_alter[col_0_17] / gesamt_population) * 100
alter_18_34_anteil = (pivot_alter[col_18_34] / gesamt_population) * 100
alter_65_plus_anteil = (pivot_alter[col_65_plus] / gesamt_population) * 100

# Erstellung des finalen Datensatzes für die statistische Analyse
df_korrelation = pd.DataFrame({
    'Migration_Anteil_%': migration_anteil,
    'Alter_0_17_Anteil_%': alter_0_17_anteil,
    'Alter_18_34_Anteil_%': alter_18_34_anteil,
    'Alter_65_plus_Anteil_%': alter_65_plus_anteil
}).dropna().round(2)


# =====================================================================
# 2. FUNKTION ZUR PRÜFUNG AUF NORMALVERTEILUNG (SHAPIRO-WILK-TEST)
# =====================================================================
def check_normality(series, title):
    clean_data = series.replace([np.inf, -np.inf], np.nan).dropna()

    if len(clean_data) < 3:
        print(f" {title} Zu wenige Daten für den Test ({len(clean_data)} Werte).\n")
        return None

    stat, p_val = stats.shapiro(clean_data)
    print(f"=== Normalverteilungstest: {title} ===")
    print(f"Stichprobengröße (N): {len(clean_data)}")
    print(f"Shapiro-Wilk-Statistik (W): {stat:.4f}")
    print(f"p-Wert: {p_val:.6e}")

    if p_val > 0.05:
        print("-> Ergebnis: Die Daten sind NORMALVERTEILT (p > 0.05)")
    else:
        print("-> Ergebnis: Die Daten sind NICHT NORMALVERTEILT (p <= 0.05)")
    print("-" * 65 + "\n")
    return clean_data


# =====================================================================
# 3. DURCHFÜHRUNG DER NORMALVERTEILUNGSTESTS FÜR DIE ZIELVARIABLEN
# =====================================================================
print("📊 ERGEBNISSE DER NORMALVERTEILUNGSTESTS FÜR PROZENTUALE ANTEILE:\n")

data_mig = check_normality(df_korrelation['Migration_Anteil_%'], "Migrationsanteil (%)")
data_age_0_17 = check_normality(df_korrelation['Alter_0_17_Anteil_%'], "Anteil Altersgruppe 0–17 (%)")
data_age_18_34 = check_normality(df_korrelation['Alter_18_34_Anteil_%'], "Anteil Altersgruppe 18–34 (%)")
data_age_65 = check_normality(df_korrelation['Alter_65_plus_Anteil_%'], "Anteil Altersgruppe 65+ (%)")


# =====================================================================
# 4. VISUALISIERUNG DER VARIABLENVERTEILUNGEN (HISTOGRAMME MIT KDE)
# =====================================================================
fig, axes = plt.subplots(1, 4, figsize=(24, 5)) # Grid mit 4 Diagrammen

# --- Grafik 1: Verteilung des Migrationsanteils ---
if data_mig is not None:
    sns.histplot(data_mig, kde=True, ax=axes[0], color='#1f77b4')
    axes[0].set_title('Bevölkerung mit Migrationshintergrund (%)', fontsize=11, fontweight='bold')
    axes[0].set_xlabel('Anteil (%)')
    axes[0].set_ylabel('Häufigkeit')

# --- Grafik 2: Verteilung der Altersgruppe 0-17 ---
if data_age_0_17 is not None:
    sns.histplot(data_age_0_17, kde=True, ax=axes[1], color='#9467bd')
    axes[1].set_title('Altersgruppe 0–17 (%)', fontsize=11, fontweight='bold')
    axes[1].set_xlabel('Anteil (%)')
    axes[1].set_ylabel('')

# --- Grafik 3: Verteilung der Altersgruppe 18-34 ---
if data_age_18_34 is not None:
    sns.histplot(data_age_18_34, kde=True, ax=axes[2], color='#ff7f0e')
    axes[2].set_title('Altersgruppe 18–34 (%)', fontsize=11, fontweight='bold')
    axes[2].set_xlabel('Anteil (%)')
    axes[2].set_ylabel('')

# --- Grafik 4: Verteilung der Altersgruppe 65+ ---
if data_age_65 is not None:
    sns.histplot(data_age_65, kde=True, ax=axes[3], color='#d62728')
    axes[3].set_title('Altersgruppe 65+ (%)', fontsize=11, fontweight='bold')
    axes[3].set_xlabel('Anteil (%)')
    axes[3].set_ylabel('')

# Akademischer Gesamttitel der Abbildung
plt.suptitle('Häufigkeitsverteilungen der analysierten Strukturvariablen (Ingolstadt)', y=1.05, fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

Die Prüfung auf Normalverteilung mittels Shapiro-Wilk-Test ergab für alle untersuchten Variablen signifikante Abweichungen von der Normalverteilung (p < 0,001). Dieses Ergebnis wird durch die grafische Analyse der Histogramme bestätigt, die überwiegend rechtsschiefe und asymmetrische Verteilungen zeigen. Für die weiteren Korrelationsanalysen werden daher überwiegend nichtparametrische Verfahren (Spearman-Rangkorrelation) verwendet.

```python
# =====================================================================
# 3. STATISTISCHE KORRELATIONSANALYSE (SPEARMAN-RHO)
# =====================================================================
print("📊 SPEARMAN-RANGKORRELATIONSANALYSE (BASIEREND AUF DF_KORRELATION):\n")

# Da 'Jahr' und 'Bezirk_standart' im Index von df_korrelation liegen können,
# erstellen wir eine temporäre Kopie mit zurückgesetztem Index für die Analyse
df_analysis = df_korrelation.copy()
if 'Jahr' not in df_analysis.columns:
    df_analysis = df_analysis.reset_index()

# Definition der zu analysierenden Paare
paare = [
    ('Migration_Anteil_%', 'Alter_0_17_Anteil_%', 'Migrationsanteil vs. Anteil Kinder/Jugendliche (0–17)'),
    ('Migration_Anteil_%', 'Alter_18_34_Anteil_%', 'Migrationsanteil vs. Anteil junge Erwachsene (18–34)'),
    ('Migration_Anteil_%', 'Alter_65_plus_Anteil_%', 'Migrationsanteil vs. Anteil Senioren (65+)')
]

for x_col, y_col, label in paare:
    # Eventuelle fehlende Werte ausschließen
    temp_data = df_analysis[[x_col, y_col]].dropna()
    
    if len(temp_data) > 2:
        rho, p_val = stats.spearmanr(temp_data[x_col], temp_data[y_col])
        
        print(f"=== {label} ===")
        print(f"Anzahl der Beobachtungen (N): {len(temp_data)} (Stadtbezirke x Jahre)")
        print(f"Spearman's Rho (ρ): {rho:.4f}")
        print(f"p-Wert: {p_val:.6e}")
        
        # Interpretation der Effektstärke nach Standardkonventionen
        abs_rho = abs(rho)
        intensitaet = "schwache" if abs_rho < 0.3 else "moderate" if abs_rho < 0.5 else "starke" if abs_rho < 0.7 else "sehr starke"
        richtung = "positive" if rho > 0 else "negative"
        signifikanz = "SIGNIFIKANT (p <= 0.05)" if p_val <= 0.05 else "NICHT SIGNIFIKANT (p > 0.05)"
        
        print(f"Ergebnis: Eine {intensitaet}, {richtung} Korrelation ({signifikanz}).")
        print("-" * 65 + "\n")
```

Die statistische Analyse der Zusammenhänge zwischen dem Migrationsanteil und der Altersstruktur in den 12 Ingolstädter Stadtbezirken (Längsschnitt 2013–2025,  N=156  Beobachtungen) liefert hochsignifikante Ergebnisse, die wichtige soziodemografische Dynamiken offenbaren:
1. Migrationsanteil vs. Altersgruppe 0–17 Jahre (Kinder und Jugendliche)

Kennzahlen: Spearman’s  ρ=−0,28  ( p<0,001 , hochsignifikant).
Interpretation: Es besteht eine schwache, aber statistisch hochsignifikante negative Korrelation. 
Soziologische Einordnung: Entgegen der oft verbreiteten Annahme, dass ein höherer Migrationsanteil in den Stadtbezirken automatisch mit einem höheren Kinderanteil einhergeht, zeigt sich in Ingolstadt das Gegenteil. In Stadtbezirken mit überdurchschnittlich hohem Migrationsanteil ist der Anteil der Kinder und Jugendlichen (0–17 Jahre) tendenziell etwas geringer. Dies deutet darauf hin, dass Familien mit Kindern (unabhängig vom Migrationshintergrund) möglicherweise in bestimmte andere Stadtteile ausweichen (z. B. suburban geprägte Randbezirke mit mehr Wohnraum für Familien).

2. Migrationsanteil vs. Altersgruppe 18–34 Jahre (Junge Erwachsene / Erwerbspersonen)

Kennzahlen: Spearman’s  ρ=0,63  ( p=1,90×10−18 , extrem signifikant).
Interpretation: Es besteht eine starke und statistisch extrem signifikante positive Korrelation.
Soziologische Einordnung: Dies ist der stärkste und prägnanteste Befund der Analyse. Stadtbezirke mit einem hohen Anteil an Personen mit Migrationshintergrund sind demografisch stark durch junge Erwachsene (im Alter von 18 bis 34 Jahren) geprägt. Dies reflektiert die typischen Migrations- und Mobilitätsmuster: Junge Menschen ziehen häufig aus Gründen der Ausbildung (z. B. Studium an der TH Ingolstadt) oder des Berufseinstiegs (z. B. Beschäftigung in der Industrie, bei Audi oder Zulieferern) in urbane Zentren und bevorzugen dort gut angebundene Stadtbezirke mit kleinerem, bezahlbarem Wohnraum.

3. Migrationsanteil vs. Altersgruppe 65+ (Senioren)

Kennzahlen: Spearman’s  ρ=0,24  ( p<0,01 , statistisch signifikant).
Interpretation: Es besteht eine schwache, aber signifikante positive Korrelation.
Soziologische Einordnung: In Stadtbezirken mit höherem Migrationsanteil liegt auch der Anteil der älteren Bevölkerung leicht höher. Dies lässt sich durch den sogenannten „Aging in Place“-Effekt der früheren Gastarbeitergenerationen (z. B. Arbeitsmigration der 1960er bis 1980er Jahre) erklären, die in ihren angestammten Wohnvierteln in Ingolstadt alt geworden sind. Dennoch ist dieser Zusammenhang deutlich schwächer ausgeprägt als der starke Zuzug der jüngeren Generation (18–34 Jahre).

### Zeitliche Dynamik und Stabilität der Korrelation (2013–2025) <a id="zeitliche_dynamik"></a> 

Die Altersgruppe der 18- bis 34-Jährigen wird zusätzlich jahresweise analysiert. Obwohl die Stichprobengröße pro Jahr lediglich N = 12 Stadtbezirke beträgt, ist die Anwendung der Spearman-Rangkorrelation methodisch vertretbar, da dieses nichtparametrische Verfahren keine Normalverteilung voraussetzt und auch für kleine Stichproben geeignet ist.

```python

# === BEHEBUNG: Sichere Kopie erstellen und 'Jahr' aus dem Index extrahieren ===
df_reg = df_korrelation.copy()

if 'Jahr' not in df_reg.columns:
    df_reg = df_reg.reset_index()

# Spaltennamen mit oder ohne Prozentzeichen zuweisen
col_migration = 'Migration_Anteil_%' if 'Migration_Anteil_%' in df_reg.columns else 'Migration_Anteil'
col_alter_18_34 = 'Alter_18_34_Anteil_%' if 'Alter_18_34_Anteil_%' in df_reg.columns else 'Alter_18_34_Anteil'

# Sicherstellen, dass die Spalte 'Jahr' jetzt definitiv vorhanden ist
if 'Jahr' not in df_reg.columns:
    raise KeyError("Die Spalte 'Jahr' wurde weder in den Spalten noch im Index gefunden. Bitte Struktur von df_korrelation prüfen.")

jahre_liste = sorted(df_reg['Jahr'].unique())
ergebnisse_nach_jahren = []

for jahn in jahre_liste:
    # Daten für das jeweilige Jahr filtern (N = 12 Stadtbezirke)
    df_jahr = df_reg[df_reg['Jahr'] == jahn]
    
    # Spearman-Rangkorrelation berechnen
    rho, p_val = stats.spearmanr(df_jahr[col_migration], df_jahr[col_alter_18_34])
    
    ergebnisse_nach_jahren.append({
        'Jahr': int(jahn),
        'Spearman_Rho': round(rho, 4),
        'p_Wert': p_val,
        'Signifikant_0.05': 'Ja' if p_val < 0.05 else 'Nein'
    })

# DataFrame mit den jährlichen Ergebnissen erstellen
df_jahres_trend = pd.DataFrame(ergebnisse_nach_jahren)

print("\n📈 KORRELATIONSTREND NACH JAHREN (Altersgruppe 18-34):")
print(df_jahres_trend.to_string(index=False))
```

Um zu überprüfen, ob der im Gesamtdatensatz (N = 156) festgestellte starke Zusammenhang zwischen dem Migrationsanteil und der Altersgruppe der 18- bis 34-Jährigen über den gesamten Untersuchungszeitraum hinweg stabil bleibt oder durch einzelne Jahre beeinflusst wird, wurde zusätzlich eine jahrweise Spearman-Korrelationsanalyse durchgeführt.

Obwohl die Stichprobengröße pro Jahr mit N = 12 Stadtbezirken vergleichsweise klein ist, zeigen die Ergebnisse eine bemerkenswerte Stabilität. Für jedes einzelne Jahr im Zeitraum von 2013 bis 2025 wurde ein statistisch signifikanter positiver Zusammenhang zwischen dem Migrationsanteil und dem Anteil der 18- bis 34-Jährigen festgestellt (p < 0,05). Der Spearman-Korrelationskoeffizient variiert dabei zwischen ρ = 0,587 im Jahr 2016 und ρ = 0,816 im Jahr 2023.

Zwischen den Jahren 2016 und 2018 ist ein vorübergehender Rückgang der Korrelationsstärke zu beobachten. Trotz dieser Schwankung bleibt der Zusammenhang in jedem einzelnen Untersuchungsjahr statistisch signifikant. Ab dem Jahr 2019 steigt die Korrelation wieder an und bewegt sich überwiegend auf einem hohen Niveau (ρ > 0,70), wobei im Jahr 2023 der höchste Korrelationskoeffizient des gesamten Untersuchungszeitraums erreicht wird.

Insgesamt sprechen die Ergebnisse dafür, dass der positive Zusammenhang zwischen dem Migrationsanteil und dem Anteil der 18- bis 34-Jährigen über den gesamten Untersuchungszeitraum hinweg weitgehend stabil ist. Der im Gesamtdatensatz beobachtete Zusammenhang wird somit nicht durch einzelne Ausreißerjahre verursacht, sondern zeigt sich konsistent über alle untersuchten Jahre hinweg.

Mögliche Erklärungsansätze könnten in ähnlichen Wohnstandortpräferenzen junger Erwachsener und von Personen mit Migrationshintergrund sowie in der räumlichen Konzentration von Ausbildungs-, Studien- und Arbeitsmöglichkeiten innerhalb der Stadt liegen. Diese Zusammenhänge können jedoch auf Grundlage der vorliegenden Korrelationsanalyse nicht kausal erklärt werden und stellen daher mögliche Interpretationen der beobachteten Ergebnisse dar.

### Karten <a id="karten"></a> 

#### Bevölkerungsstruktur nach Staatsangehörigkeit in den Stadtbezirken (2025)

```python
import os
import geopandas as gpd
import folium
from folium.features import DivIcon

# =====================================================================
# 1. LADEN UND BEREINIGEN DER GEODATEN (SHAPEFILE)
# =====================================================================
shape_dir = r"C:\0_DA\Source\Projekt\Shape-Dateien_der_Stadtbezirke_und_Unterbezirke"
shapefile_path = os.path.join(shape_dir, "Bezirke_Polygon.shp")

# Stadtbezirksgrenzen laden
gdf_bezirke = gpd.read_file(shapefile_path)

# Geografischen Mittelpunkt von Ingolstadt für die Kartenausrichtung bestimmen
centroid_projected = gdf_bezirke.geometry.centroid
center_point = gpd.GeoSeries(centroid_projected, crs=gdf_bezirke.crs).to_crs(epsg=4326).iloc[0]
center_lat = center_point.y
center_lon = center_point.x

# Projektion in das WGS84-Koordinatensystem konvertieren (Standard für Folium)
gdf_bezirke = gdf_bezirke.to_crs(epsg=4326)

# Zuordnung zur Synchronisierung der Namen aus dem Shapefile mit Ihrem Statistik-Datenrahmen
mapping_dict = {
    'Friedrichshofen-Hollerst.': 'Friedr.hof-Hollerst.',
    'Münchener Straße': 'Münchener Straße'
}
gdf_bezirke['bezname_corrected'] = gdf_bezirke['bezname'].replace(mapping_dict)

# =====================================================================
# 2. DATENAUFBEREITUNG AUS IHREM DATENSATZ (JAHR 2025)
# =====================================================================
# Schritt 1: Nur das Jahr 2025 aus Ihren drei Gruppen filtern
df_2025 = df_status[df_status['Jahr'] == 2025].copy()

# Schritt 2: Tabelle in eine flache Ansicht umwandeln (Zeile = Bezirk, Spalten = Gruppen)
df_pivot_2025 = df_2025.pivot_table(
    index='Bezirk_standart', 
    columns='Bezeichnung', 
    values='Wert', 
    aggfunc='sum'
).fillna(0)

if 'Unbekannter Bezirk' in df_pivot_2025.index:
    df_pivot_2025 = df_pivot_2025.drop('Unbekannter Bezirk')

# Schritt 3: Absolute Werte in Prozentanteile pro Bezirk für das Jahr 2025 umrechnen
df_pct_2025 = df_pivot_2025.div(df_pivot_2025.sum(axis=1), axis=0) * 100
df_pct_2025 = df_pct_2025.reset_index()

# Schritt 4: Geodaten der Bezirke mit den statistischen Daten verknüpfen (Merge)
gdf_merged = gdf_bezirke.merge(
    df_pct_2025, 
    left_on='bezname_corrected', 
    right_on='Bezirk_standart', 
    how='inner'
)

# =====================================================================
# 3. INITIALISIERUNG DER INTERAKTIVEN KARTE UND HINTERGRUND
# =====================================================================
# Leere Karte ohne standardmäßig umschaltbare Kacheln erstellen (tiles=None)
m = folium.Map(
    location=[center_lat, center_lon], 
    zoom_start=12, 
    tiles=None
)

# Einen PERMANENTEN Hintergrund-Layer (ohne Beschriftungen) hinzufügen, der immer sichtbar bleibt
folium.TileLayer(
    tiles="cartodbpositronnolabels", 
    name="Hintergrundkarte", 
    control=False  # Der Benutzer kann diesen Layer nicht versehentlich ausschalten
).add_to(m)

# =====================================================================
# 4. CHOROPLETH-LAYER FÜR DIE DREI BEVÖLKERUNGSGRUPPEN ERSTELLEN
# =====================================================================

# --- Layer 1: Ausländer (Farbskala: Gelb bis Rot) ---
folium.Choropleth(
    geo_data=gdf_merged,
    name="Anteil Ausländer (2025, in %)",
    data=gdf_merged,
    columns=['bezname_corrected', 'Ausländer'],
    key_on='feature.properties.bezname_corrected',
    fill_color='YlOrRd',
    fill_opacity=0.6,          # Deckkraft der Füllung, damit Straßen durchscheinen
    line_opacity=0.7,          # Bezirksgrenzen
    line_color='white',
    line_weight=1.5,
    highlight=True,
    legend_name="Ausländer (%)",
    overlay=False              # Aktiviert den Layer-Wechsel im Radio-Button-Modus (nur ein Layer gleichzeitig aktiv)
).add_to(m)

# --- Layer 2: Deutsche mit Migrationshintergrund (Farbskala: Blau-Violett) ---
folium.Choropleth(
    geo_data=gdf_merged,
    name="Deutsche mit Migrationshintergrund (2025, in %)",
    data=gdf_merged,
    columns=['bezname_corrected', 'Deutsche mit Migrationshintergrund'],
    key_on='feature.properties.bezname_corrected',
    fill_color='PuBu',
    fill_opacity=0.6,
    line_opacity=0.7,
    line_color='white',
    line_weight=1.5,
    highlight=True,
    legend_name="Deutsche mit Migrationshintergrund (%)",
    overlay=False
).add_to(m)

# --- Layer 3: Deutsche ohne Migrationshintergrund (Farbskala: Gelb-Grün) ---
folium.Choropleth(
    geo_data=gdf_merged,
    name="Deutsche ohne Migrationshintergrund (2025, in %)",
    data=gdf_merged,
    columns=['bezname_corrected', 'Deutsche ohne Migrationshintergrund'],
    key_on='feature.properties.bezname_corrected',
    fill_color='YlGn',
    fill_opacity=0.6,
    line_opacity=0.7,
    line_color='white',
    line_weight=1.5,
    highlight=True,
    legend_name="Deutsche ohne Migrationshintergrund (%)",
    overlay=False
).add_to(m)

# =====================================================================
# 5. INTERAKTIVE HOVER-TOOLTIPS (BEIM DARÜBERFAHREN MIT DER MAUS)
# =====================================================================
# Ein unsichtbarer interaktiver Layer wird über die Karte gelegt, um beim 
# Überfahren eines Bezirks sofort die genauen Daten aller 3 Gruppen gleichzeitig anzuzeigen
style_function = lambda x: {'fillColor': '#ffffff', 'color': '#000000', 'fillOpacity': 0.0, 'weight': 0.1}
highlight_function = lambda x: {'fillColor': '#000000', 'color': '#000000', 'fillOpacity': 0.05, 'weight': 1}

info_layer = folium.GeoJson(
    gdf_merged,
    style_function=style_function,
    control=False, # Dieser Layer arbeitet im Hintergrund und ist immer aktiv
    highlight_function=highlight_function,
    tooltip=folium.GeoJsonTooltip(
        fields=['bezname_corrected', 'Ausländer', 'Deutsche mit Migrationshintergrund', 'Deutsche ohne Migrationshintergrund'],
        aliases=['Bezirk:', 'Ausländer (%):', 'Deut. mit Migration (%):', 'Deut. ohne Migration (%):'],
        style=("background-color: white; color: #333333; font-family: sans-serif; font-size: 12px; padding: 10px; border-radius: 3px; box-shadow: 3px 3px 10px rgba(0,0,0,0.2);"),
        localize=True
    )
)
info_layer.add_to(m)

# =====================================================================
# 6. STRASSENNAMEN, FLÜSSE UND BEZIRKSNAMEN ÜBER ALLE LAYER LEGEN
# =====================================================================

# A) Straßennamen und geografische Objekte (CartoDB Labels)
# Es wird ein spezieller Darstellungsbereich (CustomPane) mit hohem z-index (625) erstellt,
# damit der Browser die Beschriftungen von Straßen und Flüssen ÜBER der farbigen Füllung zeichnet.
folium.map.CustomPane("labels_pane", z_index=625, pointer_events=False).add_to(m)

folium.TileLayer(
    tiles="cartodbpositrononlylabels", 
    pane="labels_pane",
    name="Labels",
    control=False  # Dieser Layer ist IMMER aktiv und kann vom Benutzer nicht ausgeschaltet werden
).add_to(m)

# B) Große und deutliche Bezirksnamen direkt auf der Karte platzieren
for idx, row in gdf_merged.iterrows():
    # Visuellen Mittelpunkt des aktuellen Polygons ermitteln
    centroid = row.geometry.centroid
    
    # Abkürzung zur besseren Lesbarkeit vornehmen
    name_display = row['bezname_corrected']
    if name_display == "Friedr.hof-Hollerst.":
         name_display = "Friedrichshofen"
         
    # Permanenten Textmarker erstellen (ohne Nadel/Pin, nur Text)
    folium.Marker(
        location=[centroid.y, centroid.x],
        icon=DivIcon(
            icon_size=(160, 36),
            icon_anchor=(80, 18),
            html=f"""
            <div style="
                font-family: 'Helvetica Neue', Arial, sans-serif; 
                font-size: 9px; 
                font-weight: bold; 
                color: #2c3e50; 
                text-shadow: -1.5px -1.5px 0 #fff, 1.5px -1.5px 0 #fff, -1.5px 1.5px 0 #fff, 1.5px 1.5px 0 #fff;
                text-align: center;
                pointer-events: none;
            ">
                {name_display}
            </div>
            """
        )
    ).add_to(m)

# =====================================================================
# 7. LAYER-STEUERUNG HINZUFÜGEN UND SPEICHERN
# =====================================================================
# Platzierung der Layer-Steuerung (als Radio-Buttons konfiguriert)
folium.LayerControl(collapsed=False).add_to(m)

# Als interaktive HTML-Datei speichern
m.save("ingolstadt_demographics_2025.html")
m
```

Während europäische Herkunftsgruppen über den gesamten Untersuchungszeitraum hinweg weiterhin dominieren, ist ihr relativer Anteil zwischen 2013 und 2025 zurückgegangen. Gleichzeitig nahm insbesondere der Anteil asiatischer Herkunftsregionen deutlich zu. Die statistischen Analysen zeigen einen starken positiven Zusammenhang zwischen dem Anteil der Bevölkerung mit Migrationshintergrund und dem Anteil der 18- bis 34-Jährigen (Spearman ρ = 0,63; p < 0,001). Dieser Zusammenhang bleibt über den gesamten Untersuchungszeitraum von 2013 bis 2025 stabil bestehen. Gleichzeitig wurde lediglich eine schwache negative Korrelation mit der Altersgruppe der 0- bis 17-Jährigen sowie eine schwache positive Korrelation mit der Altersgruppe ab 65 Jahren festgestellt.

Darüber hinaus zeigt die Analyse eine leichte Verjüngung der Gesamtbevölkerung im Untersuchungszeitraum. Das geschätzte Durchschnittsalter sank von 44,04 Jahren im Jahr 2013 auf 43,72 Jahre im Jahr 2025.Die Ergebnisse sprechen dafür, dass Stadtbezirke mit einem höheren Anteil an Menschen mit Migrationshintergrund häufig auch einen höheren Anteil junger Erwachsener aufweisen. Damit legen die Ergebnisse einen Zusammenhang zwischen dem Migrationsanteil und einer vergleichsweise jüngeren Altersstruktur nahe. Aufgrund der verwendeten Korrelationsanalyse kann jedoch kein kausaler Zusammenhang nachgewiesen werden.

Die Aussagekraft der Analyse wird durch mehrere Faktoren eingeschränkt. Darüber hinaus standen keine detaillierten Daten zu Einkommen, Bildung, Erwerbstätigkeit oder Wanderungsbewegungen zur Verfügung. Außerdem basiert die Untersuchung auf den verfügbaren Open-Data-Datensätzen der Stadt Ingolstadt, deren Umfang und Aktualität begrenzt sind. Die Ergebnisse beschreiben statistische Zusammenhänge, erlauben jedoch keine kausalen Schlussfolgerungen.
